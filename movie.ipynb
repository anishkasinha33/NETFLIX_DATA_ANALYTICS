```python
import pandas as pd
from sqlalchemy import create_engine

# 1. Load the dataset
df = pd.read_csv("final_dataset_movies.csv")

# 2. Fix the 3 misplaced duration rows in rating
misplaced_mask = df['rating'].str.contains('min', na=False)
df.loc[misplaced_mask, 'duration'] = df.loc[misplaced_mask, 'rating']
df.loc[misplaced_mask, 'rating'] = 'Unknown'

# 3. Fill missing text values
df['director'] = df['director'].fillna('Unknown')
df['cast'] = df['cast'].fillna('Unknown')
df['country'] = df['country'].fillna('Unknown')
df['rating'] = df['rating'].fillna('Unknown')

# 4. Extract numeric durations
df['duration_min'] = df['duration'].str.extract(r'(\d+)\s*min').astype(float)
df['duration_seasons'] = df['duration'].str.extract(r'(\d+)\s*Season').astype(float)

# 5. Format dates
df['date_added'] = pd.to_datetime(df['date_added'].str.strip(), errors='coerce')

# 6. Push directly to PostgreSQL
engine = create_engine('postgresql+psycopg2://postgres:anishka@localhost:5432/netflix_db')
df.to_sql('netflix_titles', engine, if_exists='replace', index=False)

print("Table successfully exported to PostgreSQL!")
```
